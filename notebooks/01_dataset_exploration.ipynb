```python
parent_df = pd.read_sql_query("""
    SELECT tweet_id, in_response_to_tweet_id
    FROM tweets
""", conn)

print("Rows loaded:", len(parent_df))
```

```python
parent_map = dict(
    zip(
        parent_df["tweet_id"],
        parent_df["in_response_to_tweet_id"]
    )
)

print("Parent relationships:", len(parent_map))
```

```python
root_cache = {}

def get_root(tweet_id):

    # Already calculated?
    if tweet_id in root_cache:
        return root_cache[tweet_id]

    current = tweet_id
    path = []
    visited = set()

    while current is not None:

        # If we already know the root
        if current in root_cache:
            root = root_cache[current]
            break

        # Protect against cycles
        if current in visited:
            root = current
            break

        visited.add(current)
        path.append(current)

        parent = parent_map.get(current)

        # No parent → this is the root
        if pd.isna(parent):
            root = current
            break

        current = parent

    # Cache all tweets in this path
    for tweet in path:
        root_cache[tweet] = root

    return root
```

```python
# ============================================================
# STEP 6.3 — CONVERSATION-AWARE CHUNKING
# ============================================================

def create_conversation_chunks(
    df,
    max_messages=8,
    overlap=2
):
    """
    Create overlapping message windows inside each conversation.

    We keep conversation_id as metadata so retrieved chunks
    can later be grouped back together.
    """

    chunks = []

    # Sort everything chronologically
    df = df.sort_values(
        ["conversation_id", "created_at"]
    )

    for conversation_id, group in df.groupby(
        "conversation_id",
        sort=False
    ):

        group = group.reset_index(drop=True)

        # Create overlapping windows
        step = max_messages - overlap

        for start in range(
            0,
            len(group),
            step
        ):

            window = group.iloc[
                start:start + max_messages
            ]

            if len(window) == 0:
                continue

            messages = []

            for _, row in window.iterrows():

                speaker = (
                    "CUSTOMER"
                    if bool(row["inbound"])
                    else "AMAZONHELP"
                )

                text = str(row["text"]).strip()

                if not text:
                    continue

                messages.append(
                    f"[{speaker}]: {text}"
                )

            if not messages:
                continue

            chunks.append({
                "conversation_id": conversation_id,
                "start_message": start,
                "end_message": start + len(window) - 1,
                "text": "\n".join(messages),
                "message_count": len(messages)
            })

    return pd.DataFrame(chunks)
```

```python
# ============================================================
# STEP 8.1 — BUILD SUPPORT CASES
# ============================================================

def build_support_cases(df, context_messages=3):
    """
    Build retrieval documents around AmazonHelp responses.

    Each case contains:
      - recent customer/support context
      - the AmazonHelp response
      - conversation metadata
    """

    df = df.copy()

    # Make sure chronological order is correct
    df = df.sort_values(
        ["conversation_id", "created_at"]
    )

    cases = []

    for conversation_id, group in df.groupby(
        "conversation_id",
        sort=False
    ):

        group = group.reset_index(drop=True)

        for i, row in group.iterrows():

            # We only create a retrieval case when
            # AmazonHelp is responding.
            if bool(row["inbound"]):
                continue

            # Need at least one previous message
            if i == 0:
                continue

            start = max(0, i - context_messages)

            context = group.iloc[start:i + 1]

            messages = []

            for _, msg in context.iterrows():

                speaker = (
                    "CUSTOMER"
                    if bool(msg["inbound"])
                    else "AMAZONHELP"
                )

                text = str(msg["text"]).strip()

                if not text:
                    continue

                messages.append(
                    f"[{speaker}]: {text}"
                )

            if not messages:
                continue

            cases.append({
                "conversation_id": conversation_id,
                "response_tweet_id": row["tweet_id"],
                "created_at": row["created_at"],
                "text": "\n".join(messages),
                "response": str(row["text"]).strip(),
                "message_count": len(messages)
            })

    return pd.DataFrame(cases)


support_cases = build_support_cases(
    amazon_clean,
    context_messages=3
)

print("=" * 70)
print("SUPPORT CASE CORPUS")
print("=" * 70)

print("Total support cases:", len(support_cases))

print("\nColumns:")
print(support_cases.columns.tolist())

print("\nMissing values:")
print(support_cases.isnull().sum())

print("\nSample:")
print(
    support_cases.head(3)
    .to_string(index=False)
)
```

```python
# ============================================================
# STEP 8.5 — LIGHT TEXT NORMALIZATION
# ============================================================

import re

def normalize_text(text):
    text = str(text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    # Remove excessive repeated punctuation
    text = re.sub(r"([!?.,])\1{3,}", r"\1\1", text)

    return text.strip()


support_cases["text_normalized"] = (
    support_cases["text"]
    .apply(normalize_text)
)

print(
    support_cases[
        ["text", "text_normalized"]
    ].head(3).to_string(index=False)
)
```
